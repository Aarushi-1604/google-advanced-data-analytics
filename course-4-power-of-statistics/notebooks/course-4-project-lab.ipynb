### **Task 3. Hypothesis testing**

Before you conduct your hypothesis test, consider the following questions where applicable to complete your code response:

1. Recall the difference between the null hypothesis and the alternative hypotheses. What are your hypotheses for this data project?

Since the aim is to compare the engagement between groups such as verified & unverified or claim vs opinion:

Null Hypothesis: There is no significant difference in mean video count between the two groups

Alternative Hypothesis: There is a significant differencce in the mean video count between the two groups



Your goal in this step is to conduct a two-sample t-test. Recall the steps for conducting a hypothesis test:


1.   State the null hypothesis and the alternative hypothesis
2.   Choose a signficance level
3.   Find the p-value
4.   Reject or fail to reject the null hypothesis



α = 0.05 (5%)

If p-value < 0.05 → reject H₀

If p-value ≥ 0.05 → fail to reject H₀

```python
# Conduct a two-sample t-test to compare means
# Separate groups
verified = data[data['verified_status'] == 'verified']['video_view_count']
not_verified = data[data['verified_status'] == 'not verified']['video_view_count']

print("Verified count:", len(verified))
print("Not verified count:", len(not_verified))

t_stat, p_value = stats.ttest_ind(verified, not_verified, equal_var=False)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

**Question:** Based on the p-value you got above, do you reject or fail to reject the null hypothesis?


As the P-value is basically zero, we **reject the null hypothesis**

There is a statistically significant difference in mean view counts between verified and not verified users.

T-statistic is negative too. That means the first group (verified) has a LOWER mean than the second group (Not verified), implying that not verified users, on average, have a higher view counts.
